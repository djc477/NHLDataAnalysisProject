The goal today is to figure out the cap hit value of a point share. I came up with a few methods:

 * Efficient allocation: $/PS = (32 teams x cap ceiling) / total league point shares
 * High low method: (cap ceiling - cap floor) / (league max points - league min points)
 * Division method: (cap ceiling - cap floor) / (avg division winner points - avg division last place points)

```python
ceiling26 = 95500000
floor26 = 70600000
maxpoints = 121
minpoints = 58
div_win_avg = 109.5
div_loss_avg = 71.25
ceiling27 = 104000000
```

**Method 1**

I collected point share data for this season and summed up all point shares. The total was 2,481.8

```python
a_value = (32 * ceiling26) / 2481.8
print("One point share is worth " + str(round(a_value, 0)) + " dollars of 2026 cap hit!")
a_adjusted_hit = (a_value/ceiling26) * ceiling27
print("This is equal to " + str(round(a_adjusted_hit, 0)) + " dollars of 2027 cap hit.")
```

This is a reasonable outcome. Let's create a method to gut check our answers!

```python
def gut_check(hit: float):
    mcdavid_ps = 12.7
    knies_ps = 5.3
    dewar_ps = 2.2
    celebrini_ps = 9.0
    mcdavid_value = round((mcdavid_ps * hit),0)
    knies_value = round((knies_ps * hit), 0)
    dewar_value = round((dewar_ps * hit), 0)
    celebrini_value = round((celebrini_ps * hit), 0)
    print("With this model, for a one-year contract, McDavid would be worth " + str(mcdavid_value) + ", Knies would be worth " + str(knies_value) + ", Dewar would be worth " + str(dewar_value) + ", and Celebrini would be worth " + str(celebrini_value) + ". Does that sound right?")
```

```python
gut_check(a_adjusted_hit)
```

Pretty solid!
